```python
%%writefile quantum_research.py
# ==========================================
# QUANTUM INFORMATION RESEARCH PORTFOLIO
# FILE: quantum_research.py
# PROJECT: 3-QUBIT GHZ STATE WITH PHASE FLIP
# ==========================================




from qiskit import QuantumCircuit
from qiskit.transpiler import generate_preset_pass_manager
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
from IPython.display import display

def run_quantum_experiment():
#from qiskit.visualization import plot_histogram

# Step 2: Create a Quantum Circuit with 2 qubits and 3 classical bits
# The classical bits are used to store the measurement data at the end
qc = QuantumCircuit(3, 3)

# Step 3: Apply a Hadamard Gate (H) to the first qubit (qubit 0)
# This puts qubit 0 into a state of "Superposition" (50% chance of 0, 50% chance of 1)
qc.h(0)
qc.z(0)

# Step 4: Apply a Controlled-NOT Gate (CNOT)
# This hooks qubit 0 (control) up to qubit 1 (target).
# This step creates the "Quantum Entanglement" between the two particles.
qc.cx(0, 1)
qc.cx(1,2)
# Step 5: Measure both qubits and store the results in the classical bits
qc.measure([0, 1, 2], [0, 1, 2])

# Step 6: Use the local AerSimulator to test the circuit
# This mimics how a real quantum computer behaves
simulator = AerSimulator()
pm = generate_preset_pass_manager(optimization_level=1, backend=simulator)
compiled_circuit = pm.run(qc)
#compiled_circuit = transpile(qc, simulator) # Prepares the circuit for the chip

# Step 7: Run the circuit 1024 times to gather statistics
job = simulator.run(compiled_circuit, shots=1024)
result = job.result()

# Step 8: Get the results (counts of how many times 00 or 11 appeared)
counts = result.get_counts()
print("Measurement Results:", counts)
#from IPython.display import display

# Create the figure object explicitly
fig = plot_histogram(counts)

# Force Colab to output the canvas inline
display(fig)
#plot_histogram(counts)
#plt.show()
# This wrapper allows the script to be safely imported or executed directly
if __name__ == "__main__":
    run_quantum_experiment()
```

```python
%%writefile quantum_research_teleport.py

import qiskit
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
def run_quantum_experiment():
# 3 qubits: q0 (Alice's message), q1 (Alice's entangled half), q2 (Bob's entangled half)
# 2 classical bits to send the message instructions
qc_teleport = QuantumCircuit(3, 2)

# Step 1: Prepare the secret message on q0 (Flip it to |1>)
qc_teleport.x(0)
qc_teleport.barrier()

# Step 2: Create a Bell pair (entanglement) between q1 and q2
qc_teleport.h(1)
qc_teleport.cx(1, 2)
qc_teleport.barrier()

# Step 3: Alice entangles her secret message (q0) with her shared pair (q1)
qc_teleport.cx(0, 1)
qc_teleport.h(0)
qc_teleport.barrier()

# Step 4: Alice measures her two qubits into classical bits
qc_teleport.measure(0, 0)
qc_teleport.measure(1, 1)
qc_teleport.barrier()

# Step 5: Bob modifies q2 using conditional logic based on Alice's classical bits
#qc_teleport.cx(1, 2)
#qc_teleport.cz(0, 2)
# Step 5: Bob modifies q2 using your exact classical bit rules!
with qc_teleport.if_test((qc_teleport.clbits[0], 1)):
    qc_teleport.z(2)  # If bit[0] is 1, phase shift q2

with qc_teleport.if_test((qc_teleport.clbits[1], 1)):
    qc_teleport.x(2)  # If bit[1] is 1, NOT gate q2


# Verify the result by measuring Bob's qubit (q2)
qc_teleport_meas = qc_teleport.copy()
qc_teleport_meas.measure_all()

# Run Simulation
sim = AerSimulator()
counts_teleport = sim.run(qc_teleport_meas, shots=1024).result().get_counts()
print("Teleportation Results:", counts_teleport)
plot_histogram(counts_teleport)
if __name__ == "__main__":
    run_quantum_experiment()
```

```python
%%writefile quantum_research_superdense.py
import qiskit
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram


def run_quantum_experiment():

# Pick a 2-bit message you want to send Bob: '00', '01', '10', or '11'
secret_message = '10'

# 2 qubits (q0 for Alice, q1 for Bob) and 2 classical bits to hold Bob's decoded message
qc_dense = QuantumCircuit(2, 2)

# Step 1: Pre-share an entangled Bell Pair between Alice (q0) and Bob (q1)
qc_dense.h(0)
qc_dense.cx(0, 1)
qc_dense.barrier()

# Step 2: Alice encodes her classical message into her single qubit (q0)
if secret_message == '01':
    qc_dense.x(0)  # NOT gate
elif secret_message == '10':
    qc_dense.z(0)  # Phase gate
elif secret_message == '11':
    qc_dense.z(0)  # Both gates!
    qc_dense.x(0)
qc_dense.barrier()

# Alice sends her single qubit (q0) to Bob over a quantum network line.
# Step 3: Bob receives q0 and undoes the entanglement to decode it
qc_dense.cx(0, 1)
qc_dense.h(0)
qc_dense.barrier()

# Step 4: Bob measures both qubits to read the classical bits
qc_dense.measure([0, 1],[0, 1])

# Run the Simulator
sim = AerSimulator()
counts_dense = sim.run(qc_dense, shots=1024).result().get_counts()
print(f"Superdense Coding Results for '{secret_message}':", counts_dense)
plot_histogram(counts_dense)
if __name__ == "__main__":
    run_quantum_experiment()
```
