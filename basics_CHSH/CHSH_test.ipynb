```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# 1. Create a circuit with 2 qubits and 2 classical bits
chsh_qc = QuantumCircuit(2, 2)

# 2. Create a Bell Pair (Entanglement)
chsh_qc.h(0)
chsh_qc.cx(0, 1)

# 3. Rotate Bob's qubit by 45 degrees (pi/4) to test non-classical angles
chsh_qc.ry(np.pi / 4, 1)

# 4. Measure both qubits
chsh_qc.measure([0, 1], [0, 1])

# 5. Simulate the experiment 1000 times
sim = AerSimulator()
job = sim.run(chsh_qc, shots=1000)
results = job.result().get_counts()

print("CHSH Measurement Outcomes (00, 01, 10, 11):")
print(results)
```
