```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# We need 2 qubits: Qubit 0 is input, Qubit 1 is the output/target qubit
dj_qc = QuantumCircuit(2, 1)

# 1. Initialize target qubit (Qubit 1) to |1> so H-gate creates state |->
dj_qc.x(1)

# 2. Put both qubits into superposition (Creates Phase Kickback environment)
dj_qc.h(0)
dj_qc.h(1)

# 3. THE ORACLE: A balanced function box (using CX creates a balanced output)
dj_qc.barrier()
dj_qc.cx(0, 1)  # If input is 1, flip target. This makes it balanced!
dj_qc.barrier()

# 4. Final H-gate to collapse the superposition back into a readable value
dj_qc.h(0)

# 5. Measure the input qubit
dj_qc.measure(0, 0)

# 6. Simulate
sim = AerSimulator()
job = sim.run(dj_qc, shots=1)
result = job.result().get_counts()

print("\nDeutsch-Jozsa Result:")
print(result)
# IF RESULT IS '1': The function is BALANCED (Quantum speedup achieved!)
# IF RESULT IS '0': The function is CONSTANT
```

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# We need 2 qubits: Qubit 0 is input, Qubit 1 is the output/target qubit
dj_constant_qc = QuantumCircuit(2, 1)

# 1. Initialize target qubit (Qubit 1) to |1> so H-gate creates state |->
dj_constant_qc.x(1)

# 2. Put both qubits into superposition (Creates Phase Kickback environment)
dj_constant_qc.h(0)
dj_constant_qc.h(1)

# 3. THE ORACLE: A constant function box
# Notice there is NO CX gate here! The box does nothing to connect the qubits.
# This means the output is CONSTANTly zero regardless of the input.
dj_constant_qc.barrier()
# (We leave this blank to represent a constant 0 box)
dj_constant_qc.barrier()

# 4. Final H-gate to collapse the superposition back into a readable value
dj_constant_qc.h(0)

# 5. Measure the input qubit
dj_constant_qc.measure(0, 0)

# 6. Simulate
sim = AerSimulator()
job = sim.run(dj_constant_qc, shots=1)
result = job.result().get_counts()

print("Deutsch-Jozsa Constant Result:")
print(result)
# IF RESULT IS '0': The function is CONSTANT (The waves stayed in phase!)
```
