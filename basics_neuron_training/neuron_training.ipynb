```python
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# We create a quantum neuron whose angle can be changed automatically
def quantum_neuron(learning_angle):
    qc = QuantumCircuit(1, 1)

    # 1. Input Data (Imagine this is an incoming signal rotating the qubit)
    qc.ry(np.pi / 3, 0)

    # 2. The Trainable Layer (The neuron adjusts this angle to balance the system)
    qc.ry(learning_angle, 0)

    # 3. Measurement
    qc.measure(0, 0)
    return qc

# Test 1: An untrained neuron (angle is 0)
untrained_qc = quantum_neuron(0.0)

# Test 2: A perfectly trained neuron (angle learns to counteract the input)
trained_qc = quantum_neuron(2 * np.pi / 3)

sim = AerSimulator()
print("Untrained Neuron Output (Messy/Random):")
print(sim.run(untrained_qc, shots=1000).result().get_counts())

print("\nTrained Neuron Output (Locked onto 1):")
print(sim.run(trained_qc, shots=1000).result().get_counts())
```
